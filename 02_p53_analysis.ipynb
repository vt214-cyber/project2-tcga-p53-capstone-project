````python
def expression_landscape(expr):
  """
  Summarise gene expression across cancer types.
  Returns a table containing:
  - cancer_type: the cancer type
  - n: number of tumour samples
  - median_log2tpm: median gene expression
  """
  # Step 1: Keep only tumour samples
  tumour = expr[expr["group"] == "tumour"]

  # Step 2: Count the tumour samples in each cancer type
  sample_counts = tumour.groupby("cancer_type").size()

  # Step 3: Calculate the median gene expression for each cancer type
  median_expression = tumour.groupby("cancer_type")["log2_tpm"].median()

  # Step 4: Create a table with the median expression values
  landscape = median_expression.reset_index(name="median_log2tpm")

  # Step 5: Add the sample counts to the table
  landscape["n"] = landscape["cancer_type"].map(sample_counts)

  # Step 6: Sort cancer types by median expression, highest first
  landscape = landscape.sort_values("median_log2tpm", ascending=False)

  # Step 7: Reset the row numbers so they start at zero
  landscape = landscape.reset_index(drop=True)

  # Step 8: Put columns in order
  landscape = landscape[["cancer_type", "n", "median_log2tpm"]]

  # Step 9: Return the finished table
  return landscape
````

## Task 3 — `compare_groups`

Now the actual question. You run the **same test twice**, on two
different splits:

1. **Tumour vs normal**, within your cancer type — is the gene
   dysregulated in cancer at all?
2. **TP53-mutant vs wild-type**, within the tumours — **does it respond
   to p53 loss?**

### Write one function, not two

These two comparisons do the same arithmetic on different columns.
Writing `compare_tumour_normal` and `compare_tp53` as two near-identical
functions is the commonest way to lose code-quality marks here.

```python
compare_groups(expr, group_col, group_a, group_b)
```

**Takes** `expr`, a tidy expression table already filtered to the rows
you want to compare; `group_col`, the column holding the group labels;
and `group_a` and `group_b`, the two labels to compare.

**Returns** a dict with exactly these keys:

```python
{"n_a": int, "n_b": int, "median_a": float, "median_b": float,
 "difference": float, "p_value": float, "significant": bool}
```

`difference` is `median_a - median_b`. `significant` is
`p_value < 0.05`. The test is a **two-sided Mann–Whitney U**
(`scipy.stats.mannwhitneyu`), and rows with a missing `log2_tpm` or a
missing group label are dropped.

### Why Mann–Whitney and not a t-test

Expression data is not normally distributed — it is skewed, and a few
tumours express enormously more than the rest. Mann–Whitney compares
**ranks**, so it assumes no shape and one extreme tumour cannot drag the
result. That is also why you report **medians**, not means.

````python
def compare_groups(expr, group_col, group_a, group_b):
    # Step 1: Remove rows with missing expression or group labels
    clean = expr.dropna(subset=["log2_tpm", group_col])

    #Step 2: Select samples belonging to group A
    a = clean[clean[group_col] == group_a]

    # Step 3: Select samples belonging to group B
    b = clean[clean[group_col] == group_b]

    # Step 4: Count the number of samples in each group
    n_a = len(a)
    n_b = len(b)

    # Step 5: Calculate the median expression in each group
    median_a = a["log2_tpm"].median()
    median_b = b["log2_tpm"].median()

    # Step 6: Calculate the difference between the medians
    difference = median_a - median_b

    # Step 7: Run a two sided Mann Whitney U Test
    test_result = mannwhitneyu(a["log2_tpm"], b["log2_tpm"], alternative="two-sided")

    # Step 8: Extract the p value and test significance
    p_value = float(test_result.pvalue)
    significant = bool(p_value < 0.05)

    # Step 9: Return the results in a dictionary
    return{
        "n_a": n_a,
         "n_b": n_b,
         "median_a": median_a,
         "median_b": median_b,
         "difference": difference,
         "p_value": p_value,
         "significant": significant
       }
````

````python
def plot_comparison(expr, gene, cancer_type):

  # Step 1: Select samples for the correct gene
  gene_samples = expr[expr["gene"] == gene]

  # Step 2: Select samples for the correct cancer type
  mine = gene_samples[gene_samples["cancer_type"] == cancer_type]

  # Step 3: Separate normal and tumour samples
  normal = mine[mine["group"] == "normal"]
  tumours = mine[mine["group"] == "tumour"]

  # Step 4: Separate tumour samples by TP53 status
  wild_type = tumours[tumours["tp53_status"] == "WT"]
  mutant = tumours[tumours["tp53_status"] == "MUT"]

  # Step 5: Create a list to hold the plotting data
  rows = []

  # Step 6: Add normal sample expression values
  for value in normal["log2_tpm"]:
    rows.append({"plot_group": "normal", "log2_tpm": value})

  # Step 7: Add wild type sample expression values
  for value in wild_type["log2_tpm"]:
    rows.append({"plot_group": "wild_type", "log2_tpm": value})

  # Step 8: Add mutant sample expression values
  for value in mutant["log2_tpm"]:
    rows.append({"plot_group": "mutant", "log2_tpm": value})

  # Step 9: Create a DataFrame for Seaborn
  plot_data = pd.DataFrame(rows)

  # Step 10: Create one figure and one set of axes
  fig, ax = plt.subplots(figsize=(10,6))

  # Step 11: Set the order of the three groups
  group_order = ["normal", "wild_type", "mutant"]

  # Step 12: Choose the colours for each group
  group_colours = {
      "normal": "lightblue",
      "wild_type": "lightgreen",
      "mutant": "lightcoral"
  }

  # Step 13: Create the coloured boxplots
  sns.boxplot(
      data=plot_data,
      x="plot_group",
      y="log2_tpm",
      hue="plot_group",
      order=group_order,
      palette=group_colours,
      dodge=False,
      showfliers=False,
      legend=False,
      medianprops={"color": "red", "linewidth": 2},
      ax=ax
  )
  # Step 14: Add individual sample points to the boxplot
  sns.stripplot(
      data=plot_data,
      x="plot_group",
      y="log2_tpm",
      hue="plot_group",
      order=group_order,
      palette="dark:black",
      dodge=False,
      jitter=0.2,
      size=3,
      alpha=0.5,
      legend=False,
      ax=ax
  )
  # Step 15: Add the title and axis labels
  ax.set_title(f"{gene} expression in {cancer_type}", fontsize=15)
  ax.set_xlabel("Sample Groups", fontsize=12)
  ax.set_ylabel("Expression (log2(TPM + 0.001))", fontsize=12)

  # Step 16: Add descriptive labels to the x-axis
  ax.set_xticks([0, 1, 2])
  ax.set_xticklabels([
      "Normal Tissue",
      "TP53 Wild Type Tumour",
      "TP53 Mutant Tumour"
      ])
  # Step 17: Improve spacing around the graph
  fig.tight_layout()

  # Step 18: Return the figure
  return fig
````

## Pulling it together — `run_analysis`

One function that runs the whole analysis and returns the answer.
**Each key is marked separately**, so a key you cannot produce costs you
that key and nothing else.


````python
def run_analysis(expression_path, gene, cancer_type):
    # Step 1: read the expression CSV file
    raw = pd.read_csv(expression_path)

    # check the data has loaded
    print("Expression data loaded", len(raw), "rows")

    # Step 2: Select samples for the requested gene
    gene_data = raw[raw["gene"] == gene]

    # Step 3: Summarise expression across cancer types
    landscape = expression_landscape(gene_data)

    # Step 4: Find the cancer type with the highest median expression
    landscape_top = landscape.iloc[0]["cancer_type"]

    # Check result
    print("Cancer type with highest expression:", landscape_top)

    # Step 5: Select samples from the requested cancer type
    cancer_data = gene_data[gene_data["cancer_type"] == cancer_type]

    # Step 6: Count the number of samples
    n_samples = len(cancer_data)

    # Check the number of samples
    print("Number of samples:", n_samples)

    # Step 7: Compare tumour samples with normal samples
    tumour_vs_normal = compare_groups(
        cancer_data,
        "group",
        "tumour",
        "normal"
    )

    # Step 8: Select tumour samples only
    tumour_samples = cancer_data[cancer_data["group"] == "tumour"]

    # Step 9: Compate TP53 mutant and wild type tumours
    tp53 = compare_groups(
        tumour_samples,
        "tp53_status",
        "MUT",
        "WT"
    )

    # Step 10: Decide the verdict using the TP53 comparison
    if tp53["significant"] == False:
      verdict = "no change"
    elif tp53["difference"] < 0:
      verdict = "down"
    else:
      verdict = "up"

    # Step 11: Return the results in a dictionary
    return {
        "gene": gene,
        "cancer_type": cancer_type,
        "n_samples": n_samples,
        "landscape_top": landscape_top,
        "tumour_vs_normal": tumour_vs_normal,
        "tp53": tp53,
        "verdict": verdict
    }
````
